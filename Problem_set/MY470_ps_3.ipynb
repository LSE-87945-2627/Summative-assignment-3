#### The US Congress in brief

Congress has two chambers. The **Senate** has 100 senators: two per state, regardless of population. The **House of Representatives** has 435 voting members, distributed among the states by population, from 52 for California to 1 for the smallest states. The members of both chambers from one state form the state's **delegation** to Congress.

Washington, D.C. and five US territories (e.g. Puerto Rico) send one delegate each to the House. These delegates can speak and propose bills, but cannot vote on the final passage of laws, and these places have no senators. We therefore leave them out and only include the 50 states.

#### The data

We use the file `congress_members.csv`, which you can find in the `data` repository. It contains the 533 members of Congress from the 50 states as of 27 October 2026 (433 representatives, as 2 House seats are currently vacant, and 100 senators), taken from the [congress-legislators](https://github.com/unitedstates/congress-legislators) project (public domain, CC0). Each row is one member:

| Column | Content |
|:-------|:--------|
| `first_name`, `last_name` | the member's name |
| `party` | `Democrat`, `Republican`, or `Independent` |
| `state` | two-letter state code, e.g. `VT` for Vermont |
| `chamber` | `House` or `Senate` |
| `gender` | `F` or `M` |
| `birth_year` | year of birth |
| `first_year` | year in which the member first entered Congress (in either chamber) |

### Problem 2: Class for a member of Congress

Write a class that represents one member of Congress. The specifications are:
* The class is called `Legislator`.
* `__init__()` takes the parameters `first_name`, `last_name`, `party`, `state`, `chamber`, `birth_year`, and `first_year` and stores them as data attributes (the two years as integers).
* `get_name()` returns the full name, e.g. `'Bernie Sanders'`.
* `get_party()`, `get_state()`, and `get_chamber()` return the party, the state, and the chamber.
* `get_age()` returns the age the member reaches this year (the current year minus the year of birth).
* `get_years_in_office()` returns the number of years since the member first entered Congress (the current year minus `first_year`).
* Printing a `Legislator` shows the name, the first letter of the party, the state, and the chamber, e.g. `Bernie Sanders (I-VT, Senate)`.
* `legislator1 < legislator2` returns `True` if `legislator1` is more senior, i.e. entered Congress earlier. If both entered in the same year, compare their last names in alphabetical order.

#### Hints

Your class needs the special methods `__init__()`, `__str__()`, and `__lt__()` (see the week 5 lecture).

#### Marking

4 points available for the class (plus 1 point for the test case below):
* 1 point if the class is set up correctly (correct name and `__init__()` correctly implemented)
* 1 point if all get methods are correctly implemented: `get_name()` and `get_party()` and `get_state()` and `get_chamber()` and `get_age()` and `get_years_in_office()`
* 1 point if `__str__()` is correctly implemented
* 1 point if `__lt__()` is correctly implemented

````python
# TODO: Write your class for Problem 2 here

# Note: relies on `from datetime import date`, imported in the
# notebook's first cell (the assignment requires all imports at the top).


class Legislator:
    """Represent one member of the US Congress."""

    def __init__(self, first_name, last_name, party, state,
                 chamber, birth_year, first_year):
        """Initialise a legislator with their personal and political details.

        birth_year and first_year may be passed as strings (as read by the
        csv module) and are converted to integers here.
        """
        self.first_name = first_name
        self.last_name = last_name
        self.party = party
        self.state = state
        self.chamber = chamber
        # Store the years as integers so that age and seniority
        # calculations work directly on values read from the CSV.
        self.birth_year = int(birth_year)
        self.first_year = int(first_year)

    def get_name(self):
        """Return the legislator's full name, e.g. 'Bernie Sanders'."""
        return f"{self.first_name} {self.last_name}"

    def get_party(self):
        """Return the legislator's political party."""
        return self.party

    def get_state(self):
        """Return the legislator's two-letter state code."""
        return self.state

    def get_chamber(self):
        """Return the legislator's chamber ('House' or 'Senate')."""
        return self.chamber

    def get_age(self):
        """Return the age the legislator reaches this year."""
        return date.today().year - self.birth_year

    def get_years_in_office(self):
        """Return the number of years since first entering Congress."""
        return date.today().year - self.first_year

    def __str__(self):
        """Return a description such as 'Bernie Sanders (I-VT, Senate)'."""
        # Reuse the get methods rather than reading attributes directly.
        return (f"{self.get_name()} ({self.get_party()[0]}-"
                f"{self.get_state()}, {self.get_chamber()})")

    def __lt__(self, other):
        """Return True if self is more senior than other.

        A legislator is more senior if they entered Congress earlier;
        if both entered in the same year, last names are compared
        alphabetically.
        """
        # Tuples compare element by element: first_year first, then
        # last_name only if the years are equal.
        return ((self.first_year, self.last_name)
                < (other.first_year, other.last_name))
````

````python
# TODO: Read the data for Problem 3 here

# Read the data once and create one Legislator object per row
legislators = []
with open("../data/congress_members.csv", newline="") as file:
    reader = csv.reader(file)
    next(reader)  # Skip the column names
    for row in reader:
        # Unpack by column name; gender is read but not needed
        (first_name, last_name, party, state, chamber,
         _gender, birth_year, first_year) = row
        legislators.append(Legislator(first_name, last_name, party, state,
                                      chamber, birth_year, first_year))

print("Number of legislators:", len(legislators))

print("\nFirst two legislators:")
for legislator in legislators[:2]:
    print(legislator)

# sorted() uses Legislator.__lt__(): earlier first_year first,
# ties broken by last name
most_senior = sorted(legislators)[:7]
print("\nSeven most senior legislators:")
for legislator in most_senior:
    print(legislator)
````
